```python
import numpy as np 
import pandas as pd
import os 
from glob import glob
from pathlib import Path
import sys
sys.path.insert(0, "/home/fkrafft")
from  csky_gfu_tests.custom_gfu_specs import GFUDataSpecs as custom_gfu
import csky as cy
import inspect
timer = cy.timing.Timer()
time = timer.time
```

```python
remote_root= '/data/ana/PointSource/GFU/online_v001-p10/gfu_online/version-001-p09/*'
north_path = '/data/user/fkrafft/gfu_data_sky_split/north'
south_path = '/data/user/fkrafft/gfu_data_sky_split/south'
```

```python
files = glob(remote_root)
files
```

```python
data_dict = {}
sim = None

for path in files:
    path = Path(path)
    if '_data' in path.name:
        data_dict[path.name] = np.load(path)
    elif '2023_MC' in path.name:
        sim = np.load(path)
```

```python
north_sim = sim[sim['dec'] >= np.deg2rad(-8)]
south_sim = sim[sim['dec'] < np.deg2rad(-8)]
np.save(os.path.join(north_path, 'IC86_2023_MC.npy'), north_sim)
np.save(os.path.join(south_path, 'IC86_2023_MC.npy'), south_sim)
```

```python
for filename, data in data_dict.items():
    north_data = data[data['dec'] >= np.deg2rad(-8)]
    south_data = data[data['dec'] < np.deg2rad(-8)]
    np.save(os.path.join(north_path, filename), north_data)
    np.save(os.path.join(south_path, filename), south_data)
```
